### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
arra3 = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

mu = np.mean(arra3, axis=0)
sigma = np.std(arra3, axis=0)
z = (arra3 - mu) / sigma

print("Нормалізована матриця --", z)
````

3. товар із найбільшими сумарними продажами;
4. місяць із найбільшим загальним обсягом продажів;
5. середній обсяг продажів кожного товару.

3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
# Код студента
arrM = np.random.randint(1, 21, (5, 5)) #[cite: 1]
print(arrM)

print("Сума головної діагоналі --", np.trace(arrM))
print("Сума побічної діагоналі --", np.trace(np.fliplr(arrM)))
print("Максимальний елемент кожного рядка --", np.max(arrM, axis=1))
print("Мінімальний елемент кожного стовпця --", np.min(arrM, axis=0))

mean_val = np.mean(arrM)
arrM[arrM < mean_val] = 0
print(f"Матриця після заміни елементів < {mean_val:.2f} на 0 --", arrM)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента
arrA4 = np.array([[1,  2,  1], [2,  1, -1], [3, -1,  2]])
arrb4 = np.array([9, 8, 3])

x_solve = np.linalg.solve(arrA4, arrb4)

print("Розв'язок системи --", x_solve)
print("Перевірка (A @ x) --", arrA4 @ x_solve)
print("Порівняння з вектором b --", np.allclose(arrA4 @ x_solve, arrb4))
````
